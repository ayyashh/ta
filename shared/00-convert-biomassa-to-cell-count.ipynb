```python
from pathlib import Path

import pandas as pd

folder_csv = Path(r"D:\00 Kuliah\TA\Experiment\Dataset\Data Sensor Kultivasi\csv\Biomass")
file_sumber = folder_csv / "GABUNGAN_gambar_dengan_sensor_1banding1_terdekat.csv"
file_hasil = folder_csv / "gabungan_dengan_jumlah_sel_dari_biomassa.csv"

kolom_biomassa = "sensor_biomassa_estimasi(g/L)"
kolom_gambar = "Nama_Gambar"
kolom_sel = "Jumlah Sel (sel/mL)"


if not file_sumber.is_file():
    raise FileNotFoundError(f"File sumber tidak ditemukan: {file_sumber}")

df = pd.read_csv(file_sumber, encoding="utf-8-sig")

kolom_diperlukan = {kolom_biomassa, kolom_gambar}
kolom_tidak_ada = kolom_diperlukan - set(df.columns)
if kolom_tidak_ada:
    raise KeyError(f"Kolom berikut tidak ditemukan: {sorted(kolom_tidak_ada)}")

if df[kolom_gambar].isna().any() or df[kolom_gambar].astype(str).str.strip().eq("").any():
    raise ValueError("Ada baris yang tidak memiliki Nama_Gambar; periksa pencocokan gambar.")

biomassa = pd.to_numeric(df[kolom_biomassa], errors="raise")
jumlah_sel = biomassa * 1.57e6

# Format angka bulat dengan titik pemisah ribuan, misalnya 121.847.700
df[kolom_sel] = jumlah_sel.map(lambda nilai: f"{nilai:,.0f}".replace(",", "."))

df.to_csv(file_hasil, index=False, encoding="utf-8-sig")

print(f"Jumlah baris: {len(df)}")
print(f"Kolom nama gambar: {kolom_gambar}")
print(f"File hasil disimpan di: {file_hasil}")
```
