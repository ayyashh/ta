# Konversi Optical Density (OD) ke Densitas Sel Mikroalga

Persamaan kalibrasi yang digunakan: `densitas sel (sel/mL) = (2.9371 × OD + 0.0031) × 10⁶`.

Notebook membaca `gabungan_dengan_gambar.csv` dan menambahkan kolom densitas sel ke setiap baris.

```python
from pathlib import Path

import pandas as pd

folder_csv = Path(r"D:\00 Kuliah\TA\Experiment\Dataset\Data Sensor Kultivasi\csv")
file_sumber = folder_csv / "gabungan_lab_analog_gambar_unik.csv"
file_hasil = folder_csv / "gabungan_dengan_jumlah_sel_dari_od.csv"

kolom_od = "Optical density (AU) (680 nm).Lab"
kolom_gambar = "Nama_Gambar_Terdekat"
kolom_sel = "Jumlah Sel dari OD (sel/mL)"

if not file_sumber.is_file():
    raise FileNotFoundError(f"File sumber tidak ditemukan: {file_sumber}")

df = pd.read_csv(file_sumber, encoding="utf-8-sig")

kolom_diperlukan = {kolom_od, kolom_gambar}
kolom_tidak_ada = kolom_diperlukan - set(df.columns)
if kolom_tidak_ada:
    raise KeyError(f"Kolom berikut tidak ditemukan: {sorted(kolom_tidak_ada)}")

if df[kolom_gambar].isna().any() or df[kolom_gambar].astype(str).str.strip().eq("").any():
    raise ValueError("Ada baris yang tidak memiliki Nama_Gambar; periksa pencocokan gambar.")

od = pd.to_numeric(df[kolom_od], errors="raise")
jumlah_sel = (2.9371 * od + 0.0031) * 1e6

# Format sebagai angka bulat dengan titik pemisah ribuan, misalnya 2.412.522
df[kolom_sel] = jumlah_sel.map(lambda nilai: f"{nilai:,.0f}".replace(",", "."))

df.to_csv(file_hasil, index=False, encoding="utf-8-sig")

print(f"Jumlah baris: {len(df)}")
print(f"Kolom OD: {kolom_od}")
print(f"File hasil disimpan di: {file_hasil}")
```
