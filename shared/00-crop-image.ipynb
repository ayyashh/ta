```python
from pathlib import Path
from PIL import Image

folder_input = Path(
    r"D:\00 Kuliah\TA\Experiment\Dataset\Data Gambar\Data Kultivasi Mikroalga K1-16\gambar_satu_per_sensor_unik\Session 7"
)
folder_output = folder_input.parent / "gambar_satu_per_sensor_unik_crop_200x200_session 7"

x, y = 1935, 731
width, height = 200, 200
batas_x = x + width
batas_y = y + height

file_gambar = sorted(
    path for path in folder_input.rglob("*")
    if path.is_file() and path.suffix.lower() == ".jpg"
)

berhasil = 0
dilewati = []

for file_input in file_gambar:
    file_output = folder_output / file_input.relative_to(folder_input)

    try:
        with Image.open(file_input) as gambar:
            if gambar.width < batas_x or gambar.height < batas_y:
                raise ValueError(
                    f"ukuran terlalu kecil ({gambar.width}x{gambar.height})"
                )

            hasil_crop = gambar.crop((x, y, batas_x, batas_y))
            file_output.parent.mkdir(parents=True, exist_ok=True)
            hasil_crop.save(file_output, quality=95)

        berhasil += 1

    except (OSError, ValueError) as error:
        dilewati.append((str(file_input), str(error)))

print(f"Berhasil di-crop: {berhasil}")
print(f"Dilewati karena rusak atau terlalu kecil: {len(dilewati)}")
print(f"Folder hasil: {folder_output}")

if dilewati:
    print("\nDaftar file yang dilewati:")
    for nama_file, alasan in dilewati:
        print(f"- {nama_file}: {alasan}")
```
