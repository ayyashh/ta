```python
from pathlib import Path
import shutil

folder_sumber = Path(
    r"D:\00 Kuliah\TA\Experiment\Dataset\Data Gambar\Data Kultivasi Mikroalga K1-16\gambar_satu_per_sensor_unik\Cropped_image"
)
folder_tujuan = folder_sumber.parent / "semua_gambar_jpg"

if not folder_sumber.is_dir():
    raise FileNotFoundError(f"Folder sumber tidak ditemukan: {folder_sumber}")

folder_tujuan.mkdir(parents=True, exist_ok=True)

file_jpg = [
    path for path in folder_sumber.rglob("*")
    if path.is_file() and path.suffix.lower() == ".jpg"
]

for path in file_jpg:
    tujuan = folder_tujuan / path.name

    # Hindari menimpa file jika ada nama yang sama.
    nomor = 1
    while tujuan.exists():
        tujuan = folder_tujuan / f"{path.stem}_{nomor}{path.suffix}"
        nomor += 1

    shutil.copy2(path, tujuan)

print(f"Jumlah file JPG yang dikumpulkan: {len(file_jpg)}")
print(f"Folder hasil: {folder_tujuan}")
```
