```python
from pathlib import Path
import cv2
import numpy as np

image_dir = Path(r"D:\00 Kuliah\TA\Experiment\Dataset\Data Gambar\gambar_terpilih")
output_dir = Path(r"D:\00 Kuliah\TA\Experiment\Model\Machine learning enabled image-based novel growth monitoring system for microalgae cultivated in a customized photobioreactor\lab_npy")


ekstensi_gambar = {".jpg"}
image_files = sorted(
    path for path in image_dir.rglob("*")
    if path.is_file() and path.suffix.lower() in ekstensi_gambar
)

if not image_files:
    raise FileNotFoundError(f"Tidak ada gambar yang ditemukan di: {image_dir}")

for image_path in image_files:
    # OpenCV membaca gambar dengan urutan kanal BGR.
    image_bgr = cv2.imread(str(image_path))
    if image_bgr is None:
        raise ValueError(f"Gambar tidak dapat dibaca: {image_path}")

    # Ubah ke RGB float dengan rentang 0–1, lalu konversi ke Lab.
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    image_rgb = image_rgb.astype(np.float32) / 255.0
    image_lab = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2LAB)

    # Simpan sebagai .npy dan pertahankan struktur subfolder.
    relative_path = image_path.relative_to(image_dir)
    output_path = (output_dir / relative_path).with_suffix(".npy")
    output_path.parent.mkdir(parents=True, exist_ok=True)

    np.save(output_path, image_lab)
    print(f"Berhasil: {image_path.name} -> {output_path.name}")

print(f"Selesai mengonversi {len(image_files)} gambar.")
print(f"Hasil disimpan di: {output_dir}")
```
